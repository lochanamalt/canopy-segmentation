```python
from ultralytics import YOLO

model = YOLO('yolo11s-seg.pt')

results = model.train(
    data='yolo_annotations/data.yaml',
    imgsz=1024,
    epochs=150,
    batch=-1, # Auto-adjusts for your RTX VRAM
    close_mosaic=10,# Turns off mosaic for final sharpening
    device=0,
    retina_masks=True
)
```

```python
from IPython.display import Image

Image(filename="runs/segment/train/confusion_matrix.png", width=1000)
```

```python
model = YOLO("runs/segment/train/weights/best.pt")

# Metrics of the validation data
metrics_val_data = model.val()
print(metrics_val_data.results_dict.keys())
```

```python
print("Validation Data Precision:", metrics_val_data.results_dict["metrics/precision(M)"])
print("Validation Data Recall:", metrics_val_data.results_dict["metrics/recall(M)"])
print("Validation Data mAP 50-95:", metrics_val_data.results_dict["metrics/mAP50-95(M)"])
print("Validation Data mAP 50:", metrics_val_data.results_dict["metrics/mAP50(M)"])
print("Validation Data fitness:", metrics_val_data.results_dict["fitness"])
```
